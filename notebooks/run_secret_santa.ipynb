# Secret Santa driver

Run a program from the YAML config. The example config uses mock mode, so no email is sent.

Start Jupyter from the project root or from `notebooks/`. The cell below finds the repository, imports the package, and prints the organizer log. That log lists program metadata, validation, and every pairing. Each participant email, when sending is enabled, still contains only that giver's recipient.

```python
import sys
from pathlib import Path

root = Path.cwd()
if not (root / "src" / "secret_santa").exists() and (root.parent / "src" / "secret_santa").exists():
    root = root.parent
sys.path.insert(0, str(root / "src"))

from secret_santa import run_program

result = run_program(root / "config.yaml.example")
print(f"Assigned {len(result.assignments)} participants.")
print(f"Removed {len(result.removed)} infeasible participants.")
```
